```python
import pandas as pd
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
data = {
    'Action': [5, 4, 1, 2, 5, 1],
    'Comedy': [3, 5, 2, 1, 4, 2],
    'Drama':  [4, 2, 5, 4, 1, 5],
    'SciFi':  [5, 4, 2, 1, 5, 1],
    'Like':   [1, 1, 0, 0, 1, 0]   # Label
}

df = pd.DataFrame(data)
df
```

```python
df.corr()
```

```python
X = df.drop('Like', axis=1)   # Features
y = df['Like']               # Label
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)
```

```python
y_pred = knn.predict(X_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Model Accuracy:", accuracy)
```

```python
new_user = np.array([[4, 3, 4, 5]])  # Action, Comedy, Drama, SciFi

recommendation = knn.predict(new_user)

if recommendation[0] == 1:
    print("Movie Recommended 👍")
else:
    print("Movie Not Recommended 👎")
```
